```python
class Book:
    def __init__(self, title=None, author=None, isbn=None, price=None):
        """
        Constructor for Book class. Initializes attributes with given values or defaults to None.
        """
        self.title = title
        self.author = author
        self.isbn = isbn
        self.price = price

    def display_info(self):
        """Displays book details."""
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"ISBN: {self.isbn}")
        print(f"Price: ${self.price}")

    def apply_discount(self, discount_percentage):
        """Applies a discount to the book price."""
        if self.price:
            self.price -= self.price * (discount_percentage / 100)
            print(f"New price after {discount_percentage}% discount: ${self.price:.2f}")
        else:
            print("Price not set.")

if __name__ == "__main__":
    n = int(input("Enter the number of books: "))
    books = []

    for _ in range(n):
        title = input("Enter book title: ")
        author = input("Enter book author: ")
        isbn = input("Enter book ISBN: ")
        price = float(input("Enter book price: "))

        book = Book(title, author, isbn, price)
        books.append(book)

    for book in books:
        book.display_info()
        discount = float(input("Enter discount percentage: "))
        book.apply_discount(discount)
```
